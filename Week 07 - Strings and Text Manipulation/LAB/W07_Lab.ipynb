### D1. Case changes comparisons

1. Create `a = "SRV-01"` and `b = "srv-01"`.
2. Print `a == b`.
3. Print the same comparison again, with `.lower()` on both sides.

## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# D14
raw_line= " Srv-01 , used=87 , total=100 "
raw_line.strip() # to remove the leading and trailing whitespace
hostname , rest = raw_line.split(",", 1) # to split the string into a list of words and limit the number of splits to 1
hostname = hostname.strip().lower() # to convert the string to lower case
rest = rest.split(",")
key, value = rest[0].split("=") 
dict = {
    "hostname": hostname,
    key.strip() : float(value.strip()),
    "total" : float(rest[1].split("=")[1])
} # to create a dictionary from the list of words
print(hostname, dict) # to split the string into a list of words and then split the first word into a key-value pair
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15
def parse_line(line):
    """this function takes a string as input and returns a dictionary with the hostname, used and total values"""
    line = line.strip() # to remove the leading and trailing whitespace
    hostname , rest = line.split(",", 1) # to split the string into a list of words and limit the number of splits to 1
    hostname = hostname.strip().lower() # to convert the string to lower case
    rest = rest.split(",")
    key, value = rest[0].split("=") 
    dict = {
        "hostname": hostname,
        key.strip() : float(value.strip()),
        "total" : float(rest[1].split("=")[1])
    } # to create a dictionary from the list of words
    return dict
print(parse_line(" Srv-02 , used=120 , total=100 ")) # to split the string into a list of words and then split the first word into a key-value pair
````

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
# C1

with open("messy_lines.txt", "r") as file:
    for raw_line in file:
        record = parse_line(raw_line)
        usage = record["used"] / record["total"] if record["total"] else 0
        if usage >= 1.0:
            status = "OVER LIMIT"
        elif usage >= 0.9:
            status = "WARNING"
        else:
            status = "OK"
        print(f"{record["hostname"]} {status}")
````

### C2. When parsing breaks

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print
`Could not parse:` and the line, instead of crashing.

```python
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
```
